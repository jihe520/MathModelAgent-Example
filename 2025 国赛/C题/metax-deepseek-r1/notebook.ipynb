```
# 添加全局错误处理
try:
    import pandas as pd
    import numpy as np
    import seaborn as sns
    import matplotlib.pyplot as plt
    import statsmodels.api as sm
    from statsmodels.stats.outliers_influence import summary_table
    
    # 配置中文显示
    plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
    plt.rcParams['axes.unicode_minus'] = False
    
    # 加载清洗后的数据
    df = pd.read_csv('cleaned_data.csv')
    
    # 准备建模数据
    X = df[['检测孕周', '孕妇BMI']]
    y = df['Y染色体浓度']
    
    # 添加常数项
    X = sm.add_constant(X)
    
    # 建立多元线性回归模型
    model = sm.OLS(y, X).fit()
    
    # 模型总结
    print("="*50)
    print("多元线性回归模型总结")
    print("="*50)
    print(model.summary())
    
    # 提取置信区间
    st, data, ss2 = summary_table(model, alpha=0.05)
    predicted = data[:, 2]  # 预测值
    ci_low, ci_high = data[:, 4:6].T  # 95%置信区间
    
    # 可视化1：回归拟合线（按BMI分组）
    plt.figure(figsize=(12, 8))
    
    # 按BMI三分位数分组
    bmi_groups = pd.qcut(df['孕妇BMI'], 3, labels=['低BMI', '中BMI', '高BMI'])
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
    
    for i, group in enumerate(['低BMI', '中BMI', '高BMI']):
        mask = (bmi_groups == group)
        plt.scatter(df.loc[mask, '检测孕周'], y.loc[mask], 
                    alpha=0.6, label=group, color=colors[i])
        
        # 计算并绘制拟合线
        x_fit = np.linspace(df['检测孕周'].min(), df['检测孕周'].max(), 100)
        y_fit = model.params['const'] + model.params['检测孕周'] * x_fit + \
                model.params['孕妇BMI'] * df.loc[mask, '孕妇BMI'].median()
        plt.plot(x_fit, y_fit, color=colors[i], linewidth=2.5)
    
    plt.xlabel('孕周')
    plt.ylabel('Y染色体浓度')
    plt.title('Y染色体浓度 vs 孕周 (按BMI分组)', fontsize=14)
    plt.legend(title='BMI分组')
    plt.grid(alpha=0.2)
    plt.savefig('regression_fit_by_bmi.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # 可视化2：残差分布图
    plt.figure(figsize=(10, 6))
    residuals = model.resid
    sns.histplot(residuals, kde=True, color='purple', bins=30)
    plt.axvline(x=0, color='r', linestyle='--', alpha=0.7)
    plt.xlabel('残差')
    plt.ylabel('频数')
    plt.title('残差分布图')
    plt.savefig('residual_distribution.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # 可视化3：预测值 vs 实际值（带置信区间）
    plt.figure(figsize=(10, 8))
    plt.scatter(predicted, y, alpha=0.6, color='navy')
    plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', alpha=0.7)  # 对角线
    
    # 添加置信区间带
    sorted_idx = np.argsort(predicted)
    plt.fill_between(predicted[sorted_idx], ci_low[sorted_idx], ci_high[sorted_idx], 
                     color='gray', alpha=0.2, label='95% 置信区间')
    
    plt.xlabel('预测值')
    plt.ylabel('实际值')
    plt.title('预测值 vs 实际值 (带置信区间)', fontsize=14)
    plt.legend()
    plt.grid(alpha=0.2)
    plt.savefig('predicted_vs_actual.png', dpi=300, bbox_inches='tight')
    plt.close()
    
    # 输出模型结论
    print("\n" + "="*50)
    print("模型结论：")
    print("="*50)
    print(f"1. 回归模型: Y染色体浓度 = {model.params['const']:.4f} + "
          f"{model.params['检测孕周']:.4f}*孕周 + {model.params['孕妇BMI']:.4f}*BMI")
    print(f"2. 模型整体显著性 (F检验): p值 = {model.f_pvalue:.4f}" + 
          (" < 0.05，模型整体显著" if model.f_pvalue < 0.05 else " > 0.05，模型不显著"))
    print(f"3. 孕周系数显著性 (t检验): p值 = {model.pvalues['检测孕周']:.4f}" + 
          (" < 0.05，显著" if model.pvalues['检测孕周'] < 0.05 else " > 0.05，不显著"))
    print(f"4. BMI系数显著性 (t检验): p值 = {model.pvalues['孕妇BMI']:.4f}" + 
          (" < 0.05，显著" if model.pvalues['孕妇BMI'] < 0.05 else " > 0.05，不显著"))
    print(f"5. 模型解释方差 (R²): {model.rsquared:.4f}")
    
    print("\n可视化图表已保存至当前目录")
except Exception as e:
    print(f"❌ 执行错误: {e}")
    print("📍 错误详情:")
    import traceback
    traceback.print_exc()
    
    # 尝试提供错误解决建议
    if "NoneType" in str(e):
        print("💡 建议: 检查变量是否为None")
    elif "KeyError" in str(e):
        print("💡 建议: 检查字典键或DataFrame列名是否存在")
    elif "IndexError" in str(e):
        print("💡 建议: 检查列表或数组索引是否越界")
    
    raise  # 重新抛出异常以便调试
```

```
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import summary_table

# 配置中文显示
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 加载清洗后的数据
df = pd.read_csv('cleaned_data.csv')

print("="*50)
print("详细数据诊断报告")
print("="*50)

# 1. 检查关键列数据类型
print("原始数据类型:")
print(df[['检测孕周', '孕妇BMI', 'Y染色体浓度']].dtypes)

# 2. 检查缺失值
print("\n原始缺失值统计:")
print(df[['检测孕周', '孕妇BMI', 'Y染色体浓度']].isna().sum())

# 3. 尝试转换为数值
for col in ['检测孕周', '孕妇BMI', 'Y染色体浓度']:
    df[f'{col}_numeric'] = pd.to_numeric(df[col], errors='coerce')

# 4. 检查转换后的缺失值
print("\n转换后缺失值统计:")
print(df[['检测孕周_numeric', '孕妇BMI_numeric', 'Y染色体浓度_numeric']].isna().sum())

# 5. 删除无效行
df_clean = df.dropna(subset=['检测孕周_numeric', '孕妇BMI_numeric', 'Y染色体浓度_numeric'])
print(f"\n有效样本数量: {len(df_clean)}")

# 6. 回退方案：如果没有BMI数据，使用身高体重计算BMI
if len(df_clean) == 0:
    print("⚠️ 警告：所有BMI数据无效，尝试使用身高体重计算BMI")
    
    # 确保身高体重列存在
    if '身高(cm)' in df.columns and '体重(kg)' in df.columns:
        # 转换为数值
        df['身高_numeric'] = pd.to_numeric(df['身高(cm)'], errors='coerce')
        df['体重_numeric'] = pd.to_numeric(df['体重(kg)'], errors='coerce')
        
        # 计算BMI: weight(kg)/(height(m)^2)
        df['BMI_calculated'] = df['体重_numeric'] / (df['身高_numeric']/100)**2
        
        # 准备新数据集
        df_clean = df.dropna(subset=['检测孕周_numeric', 'BMI_calculated', 'Y染色体浓度_numeric'])
        print(f"使用计算BMI的有效样本数: {len(df_clean)}")
        
        # 更新列名
        bmi_col = 'BMI_calculated'
    else:
        print("❌ 错误：缺少身高体重列，无法计算BMI")
        raise ValueError("所有数据无效且无法计算BMI")
else:
    bmi_col = '孕妇BMI_numeric'

# 如果仍然没有数据，报告错误
if len(df_clean) == 0:
    print("❌ 错误：没有有效数据可用于建模")
    print("请检查原始数据文件中的以下列：")
    print("- 检测孕周")
    print("- 孕妇BMI 或 身高(cm)/体重(kg)")
    print("- Y染色体浓度")
    raise ValueError("没有有效数据可用于建模")

# 准备建模数据
X = df_clean[['检测孕周_numeric', bmi_col]]
y = df_clean['Y染色体浓度_numeric']

# 重命名列以用于输出
X.columns = ['检测孕周', '孕妇BMI']

# 添加常数项
X = sm.add_constant(X)

# 建立多元线性回归模型
model = sm.OLS(y, X).fit()

# 模型总结
print("="*50)
print("多元线性回归模型总结")
print("="*50)
print(model.summary())

# 提取置信区间
st, data, ss2 = summary_table(model, alpha=0.05)
predicted = data[:, 2]  # 预测值
ci_low, ci_high = data[:, 4:6].T  # 95%置信区间

# 可视化1：回归拟合线（按BMI分组）
plt.figure(figsize=(12, 8))

# 按BMI三分位数分组
bmi_groups = pd.qcut(df_clean[bmi_col], 3, labels=['低BMI', '中BMI', '高BMI'])
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

for i, group in enumerate(['低BMI', '中BMI', '高BMI']):
    mask = (bmi_groups == group)
    plt.scatter(df_clean.loc[mask, '检测孕周_numeric'], y.loc[mask], 
                alpha=0.6, label=group, color=colors[i])
    
    # 计算并绘制拟合线
    x_fit = np.linspace(df_clean['检测孕周_numeric'].min(), df_clean['检测孕周_numeric'].max(), 100)
    y_fit = model.params['const'] + model.params['检测孕周'] * x_fit + \
            model.params['孕妇BMI'] * df_clean.loc[mask, bmi_col].median()
    plt.plot(x_fit, y_fit, color=colors[i], linewidth=2.5)

plt.xlabel('孕周')
plt.ylabel('Y染色体浓度')
plt.title('Y染色体浓度 vs 孕周 (按BMI分组)', fontsize=14)
plt.legend(title='BMI分组')
plt.grid(alpha=0.2)
plt.savefig('regression_fit_by_bmi.png', dpi=300, bbox_inches='tight')
plt.close()

# 可视化2：残差分布图
plt.figure(figsize=(10, 6))
residuals = model.resid
sns.histplot(residuals, kde=True, color='purple', bins=30)
plt.axvline(x=0, color='r', linestyle='--', alpha=0.7)
plt.xlabel('残差')
plt.ylabel('频数')
plt.title('残差分布图')
plt.savefig('residual_distribution.png', dpi=300, bbox_inches='tight')
plt.close()

# 可视化3：预测值 vs 实际值（带置信区间）
plt.figure(figsize=(10, 8))
plt.scatter(predicted, y, alpha=0.6, color='navy')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', alpha=0.7)  # 对角线

# 添加置信区间带
sorted_idx = np.argsort(predicted)
plt.fill_between(predicted[sorted_idx], ci_low[sorted_idx], ci_high[sorted_idx], 
                 color='gray', alpha=0.2, label='95% 置信区间')

plt.xlabel('预测值')
plt.ylabel('实际值')
plt.title('预测值 vs 实际值 (带置信区间)', fontsize=14)
plt.legend()
plt.grid(alpha=0.2)
plt.savefig('predicted_vs_actual.png', dpi=300, bbox_inches='tight')
plt.close()

# 输出模型结论
print("\n" + "="*50)
print("模型结论：")
print("="*50)
print(f"1. 回归模型: Y染色体浓度 = {model.params['const']:.6f} + "
      f"{model.params['检测孕周']:.6f}*孕周 + {model.params['孕妇BMI']:.6f}*BMI")
print(f"2. 模型整体显著性 (F检验): p值 = {model.f_pvalue:.4f}" + 
      (" < 0.05，模型整体显著" if model.f_pvalue < 0.05 else " > 0.05，模型不显著"))
print(f"3. 孕周系数显著性 (t检验): p值 = {model.pvalues['检测孕周']:.4f}" + 
      (" < 0.05，显著" if model.pvalues['检测孕周'] < 0.05 else " > 0.05，不显著"))
print(f"4. BMI系数显著性 (t检验): p值 = {model.pvalues['孕妇BMI']:.4f}" + 
      (" < 0.05，显著" if model.pvalues['孕妇BMI'] < 0.05 else " > 0.05，不显著"))
print(f"5. 模型解释方差 (R²): {model.rsquared:.4f}")

print("\n可视化图表已保存至当前目录")

# 保存诊断报告
with open('data_diagnostic_report.txt', 'w') as f:
    f.write("="*50 + "\n")
    f.write("数据诊断报告\n")
    f.write("="*50 + "\n\n")
    f.write(f"原始样本数: {len(df)}\n")
    f.write(f"有效样本数: {len(df_clean)}\n\n")
    f.write("原始数据类型:\n")
    f.write(str(df[['检测孕周', '孕妇BMI', 'Y染色体浓度']].dtypes) + "\n\n")
    f.write("转换后数据类型:\n")
    try:
        f.write(str(df_clean[['检测孕周_numeric', bmi_col, 'Y染色体浓度_numeric']].dtypes) + "\n")
    except:
        f.write("数据类型转换失败\n")
    
    if 'BMI_calculated' in df_clean.columns:
        f.write("\n⚠️ 使用了计算BMI值\n")
```

```
import pandas as pd
import numpy as np
import re
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.api as sm

# 重新加载原始数据
df = pd.read_excel('附件.xlsx')

# 修复孕周列：提取数值
print("修复'检测孕周'列...")
df['孕周_修正'] = df['检测孕周'].apply(
    lambda x: float(re.search(r'\d+\.?\d*', str(x)).group()) 
    if pd.notnull(x) and re.search(r'\d+\.?\d*', str(x)) 
    else np.nan
)

print(f"有效孕周数量: {df['孕周_修正'].notna().sum()}/{len(df)}")

# 准备分析数据
df_analysis = df[['孕周_修正', '孕妇BMI', 'Y染色体浓度']].dropna()
df_analysis.columns = ['孕周', 'BMI', 'Y染色体浓度']
print(f"可用样本量: {len(df_analysis)}")

# 问题2：BMI分组分析
print("\n===== 问题2分析 =====")

# BMI分组
bmi_bins = [0, 20, 28, 32, 36, 40, 100]
bmi_labels = ['<20', '20-28', '28-32', '32-36', '36-40', '>40']
df_analysis['BMI分组'] = pd.cut(df_analysis['BMI'], bins=bmi_bins, labels=bmi_labels)

# 计算达标孕周
group_data = []
for group in bmi_labels:
    group_df = df_analysis[df_analysis['BMI分组'] == group]
    if len(group_df) > 0:
        # 找到达到4%浓度的最小孕周
        qualified = group_df[group_df['Y染色体浓度'] >= 4]
        min_week = qualified['孕周'].min() if not qualified.empty else np.nan
        
        group_data.append({
            'BMI分组': group, 
            '样本量': len(group_df),
            '达标样本数': len(qualified),
            '达标比例': len(qualified) / len(group_df) if len(group_df) > 0 else 0,
            '最佳NIPT时点': min_week
        })

# 转换为DataFrame
group_df = pd.DataFrame(group_data)
print("BMI分组分析结果:")
print(group_df)

# 可视化
plt.figure(figsize=(12, 6))

# 样本分布
plt.subplot(121)
sns.barplot(x='BMI分组', y='样本量', data=group_df, palette='viridis')
plt.title('各BMI组样本分布')

# 最佳时点
plt.subplot(122)
# 过滤掉无数据的组
plot_df = group_df.dropna(subset=['最佳NIPT时点'])
if not plot_df.empty:
    sns.pointplot(x='BMI分组', y='最佳NIPT时点', data=plot_df, join=True, color='red')
    plt.title('最佳NIPT检测时点')
else:
    plt.text(0.5, 0.5, "无达标数据", ha='center')
    
plt.tight_layout()
plt.savefig('bmi_group_analysis.png', dpi=300)
plt.close()

# 问题3：综合因素分析
print("\n===== 问题3分析 =====")
# 添加更多特征（移除身高体重避免共线性）
df_analysis['年龄'] = df['年龄']

# 删除缺失值
df_analysis = df_analysis[['孕周', 'BMI', 'Y染色体浓度', '年龄']].dropna()
print(f"综合分析样本量: {len(df_analysis)}")

# 逻辑回归模型 - 添加正则化解决共线性
X = df_analysis[['BMI', '年龄']]
y = (df_analysis['Y染色体浓度'] >= 4).astype(int)  # 达标比例

# 添加常数项
X = sm.add_constant(X)

# 使用带正则化的逻辑回归
logit_model = sm.Logit(y, X)
# 使用牛顿法求解，添加ridge正则化
logit_result = logit_model.fit_regularized(alpha=0.1, method='l1')
print("逻辑回归参数:")
print(logit_result.params)

# 计算达标概率
X['达标概率'] = logit_result.predict(X)

# 根据BMI分组计算平均达标概率
bmi_group_results = []
for group in bmi_labels:
    group_X = X[X.index.isin(df_analysis[df_analysis['BMI分组'] == group].index)]
    if not group_X.empty:
        mean_prob = group_X['达标概率'].mean()
        bmi_group_results.append({
            'BMI分组': group,
            '平均达标概率': mean_prob
        })

bmi_prob_df = pd.DataFrame(bmi_group_results)

# 保存结果
with open('group_analysis_results.txt', 'w', encoding='utf-8') as f:
    f.write("BMI分组分析结果:\n")
    f.write(group_df.to_string() + "\n\n")
    f.write("逻辑回归参数:\n")
    f.write(str(logit_result.params) + "\n\n")
    f.write("按BMI分组的平均达标概率:\n")
    f.write(bmi_prob_df.to_string())

print("分析完成！结果已保存")
print(f"分组分析图: bmi_group_analysis.png")
print(f"详细结果: group_analysis_results.txt")
```

```
# 添加全局错误处理
try:
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.model_selection import train_test_split, GridSearchCV
    from sklearn.metrics import roc_curve, auc, confusion_matrix, ConfusionMatrixDisplay
    
    # 设置科学出版级别的可视化风格
    plt.rcParams.update({
        "font.size": 12,
        "font.family": "sans-serif",
        "axes.spines.right": False,
        "axes.spines.top": False,
        "figure.figsize": (10, 6),
        "figure.dpi": 300
    })
    
    # 加载数据
    df = pd.read_excel('附件.xlsx')
    
    # 筛选女胎样本 (Y染色体浓度<4%)
    df_female = df[df['Y染色体浓度'] < 4].copy()
    print(f"女胎样本数量: {len(df_female)}")
    
    # 创建目标变量
    conditions = [
        (df_female['21－三体结果'] == '高风险') |
        (df_female['18－三体结果'] == '高风险') |
        (df_female['13－三体结果'] == '高风险')
    ]
    df_female['异常'] = np.where(np.any(conditions, axis=0), 1, 0)
    print(f"异常样本比例: {df_female['异常'].mean():.2%}")
    
    # 特征工程
    features = df_female[[
        'X染色体Z值', 
        '21三体Z值', '18三体Z值', '13三体Z值',
        'GC1', 'GC2', 'GC3', 'GC4', 'GC5', 'GC6',
        '总读段数', '唯一比对比', '重复比',
        '孕妇BMI'
    ]]
    
    target = df_female['异常']
    
    # 删除缺失值
    features = features.dropna()
    target = target.loc[features.index]
    print(f"最终样本量: {len(features)}")
    
    # 划分训练集和测试集
    X_train, X_test, y_train, y_test = train_test_split(
        features, target, test_size=0.3, random_state=42, stratify=target
    )
    
    # 随机森林模型
    rf = RandomForestClassifier(random_state=42)
    
    # 参数网格
    param_grid = {
        'n_estimators': [100, 200],
        'max_depth': [None, 10, 20],
        'min_samples_split': [2, 5],
        'class_weight': [None, 'balanced']
    }
    
    # 网格搜索
    grid_search = GridSearchCV(
        estimator=rf,
        param_grid=param_grid,
        cv=5,
        scoring='f1',
        n_jobs=-1,
        verbose=1
    )
    grid_search.fit(X_train, y_train)
    
    # 最佳模型
    best_rf = grid_search.best_estimator_
    print(f"最佳参数: {grid_search.best_params_}")
    print(f"训练集F1分数: {grid_search.best_score_:.4f}")
    
    # 预测概率
    y_proba = best_rf.predict_proba(X_test)[:, 1]
    
    # ROC曲线和最佳阈值
    fpr, tpr, thresholds = roc_curve(y_test, y_proba)
    roc_auc = auc(fpr, tpr)
    
    # 最佳阈值（最大化Youden指数）
    j_scores = tpr - fpr
    best_idx = np.argmax(j_scores)
    best_threshold = thresholds[best_idx]
    
    y_pred = (y_proba >= best_threshold).astype(int)
    
    # 特征重要性
    feature_importances = pd.Series(
        best_rf.feature_importances_,
        index=(features.columns if features is not None else [])
    ).sort_values(ascending=False)
    
    # === 可视化 ===
    
    # 1. 特征重要性
    plt.figure()
    feature_importances.plot(kind='barh', color='steelblue')
    plt.title('随机森林特征重要性')
    plt.xlabel('重要性得分')
    plt.tight_layout()
    plt.savefig('rf_feature_importance.png')
    
    # 2. ROC曲线
    plt.figure()
    plt.plot(fpr, tpr, color='darkorange', lw=2, 
             label=f'ROC曲线 (AUC = {roc_auc:.2f})')
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.scatter(fpr[best_idx], tpr[best_idx], marker='o', color='red', 
                label=f'最佳阈值={best_threshold:.2f}')
    plt.xlabel('假阳性率')
    plt.ylabel('真阳性率')
    plt.title('ROC曲线')
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.savefig('rf_roc_curve.png')
    
    # 3. 混淆矩阵
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    plt.figure()
    disp.plot(cmap='Blues', values_format='d')
    plt.title('混淆矩阵')
    plt.tight_layout()
    plt.savefig('rf_confusion_matrix.png')
    
    # 4. 概率分布
    plt.figure()
    sns.kdeplot(y_proba[y_test == 0], label='正常', fill=True, alpha=0.5)
    sns.kdeplot(y_proba[y_test == 1], label='异常', fill=True, alpha=0.5)
    plt.axvline(best_threshold, color='red', linestyle='--', label='决策阈值')
    plt.xlabel('异常概率')
    plt.ylabel('密度')
    plt.title('预测概率分布')
    plt.legend()
    plt.tight_layout()
    plt.savefig('rf_probability_distribution.png')
    
    plt.close('all')
    
    # 保存模型报告
    with open('rf_model_report.txt', 'w', encoding='utf-8') as f:
        f.write("=== 女胎异常判定模型报告 ===\n\n")
        f.write(f"样本总量: {len(df_female)}\n")
        f.write(f"异常样本比例: {target.mean():.2%}\n\n")
        
        f.write("最佳模型参数:\n")
        for k, v in grid_search.best_params_.items():
            f.write(f"{k}: {v}\n")
        
        f.write("\n特征重要性排序:\n")
        f.write(feature_importances.to_string())
        
        f.write("\n\n测试集性能:\n")
        f.write(f"- AUC: {roc_auc:.4f}\n")
        f.write(f"- 最佳阈值: {best_threshold:.4f}\n")
        
        f.write("\n可视化文件:\n")
        f.write("1. rf_feature_importance.png - 特征重要性\n")
        f.write("2. rf_roc_curve.png - ROC曲线\n")
        f.write("3. rf_confusion_matrix.png - 混淆矩阵\n")
        f.write("4. rf_probability_distribution.png - 概率分布\n")
    
    print("建模完成！结果保存至:")
    print("- rf_feature_importance.png")
    print("- rf_roc_curve.png")
    print("- rf_confusion_matrix.png")
    print("- rf_probability_distribution.png")
    print("- rf_model_report.txt")
except Exception as e:
    print(f"❌ 执行错误: {e}")
    print("📍 错误详情:")
    import traceback
    traceback.print_exc()
    
    # 尝试提供错误解决建议
    if "NoneType" in str(e):
        print("💡 建议: 检查变量是否为None")
    elif "KeyError" in str(e):
        print("💡 建议: 检查字典键或DataFrame列名是否存在")
    elif "IndexError" in str(e):
        print("💡 建议: 检查列表或数组索引是否越界")
    
    raise  # 重新抛出异常以便调试
```

```
# 添加全局错误处理
try:
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.model_selection import train_test_split, GridSearchCV
    from sklearn.metrics import roc_curve, auc, confusion_matrix, ConfusionMatrixDisplay
    
    # 设置科学出版级别的可视化风格
    plt.rcParams.update({
        "font.size": 12,
        "font.family": "sans-serif",
        "axes.spines.right": False,
        "axes.spines.top": False,
        "figure.figsize": (10, 6),
        "figure.dpi": 300
    })
    
    # 加载数据
    df = pd.read_excel('附件.xlsx')
    
    # 调试：输出所有列名以便验证
    print("数据集列名:")
    for idx, col in enumerate((df.columns if df is not None else [])):
        print(f"{idx}: {col}")
    
    # 筛选女胎样本 (Y染色体浓度<4%)
    df_female = df[df['Y染色体浓度'] < 4].copy()
    print(f"\n女胎样本数量: {len(df_female)}")
    
    # 创建目标变量 - 使用正确的长破折号字符
    conditions = [
        (df_female['21－三体结果'] == '高风险') |
        (df_female['18－三体结果'] == '高风险') |
        (df_female['13－三体结果'] == '高风险')
    ]
    df_female['异常'] = np.where(np.any(conditions, axis=0), 1, 0)
    print(f"异常样本比例: {df_female['异常'].mean():.2%}")
    
    # 特征工程
    features = df_female[[
        'X染色体Z值', 
        '21三体Z值', '18三体Z值', '13三体Z值',
        'GC1', 'GC2', 'GC3', 'GC4', 'GC5', 'GC6',
        '总读段数', '唯一比对比', '重复比',
        '孕妇BMI'
    ]]
    
    target = df_female['异常']
    
    # 删除缺失值
    features = features.dropna()
    target = target.loc[features.index]
    print(f"最终样本量: {len(features)}")
    
    # 划分训练集和测试集
    X_train, X_test, y_train, y_test = train_test_split(
        features, target, test_size=0.3, random_state=42, stratify=target
    )
    
    # 随机森林模型
    rf = RandomForestClassifier(random_state=42)
    
    # 参数网格
    param_grid = {
        'n_estimators': [100, 200],
        'max_depth': [None, 10, 20],
        'min_samples_split': [2, 5],
        'class_weight': [None, 'balanced']
    }
    
    # 网格搜索
    grid_search = GridSearchCV(
        estimator=rf,
        param_grid=param_grid,
        cv=5,
        scoring='f1',
        n_jobs=-1,
        verbose=1
    )
    grid_search.fit(X_train, y_train)
    
    # 最佳模型
    best_rf = grid_search.best_estimator_
    print(f"最佳参数: {grid_search.best_params_}")
    print(f"训练集F1分数: {grid_search.best_score_:.4f}")
    
    # 预测概率
    y_proba = best_rf.predict_proba(X_test)[:, 1]
    
    # ROC曲线和最佳阈值
    fpr, tpr, thresholds = roc_curve(y_test, y_proba)
    roc_auc = auc(fpr, tpr)
    
    # 最佳阈值（最大化Youden指数）
    j_scores = tpr - fpr
    best_idx = np.argmax(j_scores)
    best_threshold = thresholds[best_idx]
    
    y_pred = (y_proba >= best_threshold).astype(int)
    
    # 特征重要性
    feature_importances = pd.Series(
        best_rf.feature_importances_,
        index=(features.columns if features is not None else [])
    ).sort_values(ascending=False)
    
    # === 可视化 ===
    
    # 1. 特征重要性
    plt.figure()
    feature_importances.plot(kind='barh', color='steelblue')
    plt.title('随机森林特征重要性')
    plt.xlabel('重要性得分')
    plt.tight_layout()
    plt.savefig('rf_feature_importance.png')
    
    # 2. ROC曲线
    plt.figure()
    plt.plot(fpr, tpr, color='darkorange', lw=2, 
             label=f'ROC曲线 (AUC = {roc_auc:.2f})')
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.scatter(fpr[best_idx], tpr[best_idx], marker='o', color='red', 
                label=f'最佳阈值={best_threshold:.2f}')
    plt.xlabel('假阳性率')
    plt.ylabel('真阳性率')
    plt.title('ROC曲线')
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.savefig('rf_roc_curve.png')
    
    # 3. 混淆矩阵
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    plt.figure()
    disp.plot(cmap='Blues', values_format='d')
    plt.title('混淆矩阵')
    plt.tight_layout()
    plt.savefig('rf_confusion_matrix.png')
    
    # 4. 概率分布
    plt.figure()
    sns.kdeplot(y_proba[y_test == 0], label='正常', fill=True, alpha=0.5)
    sns.kdeplot(y_proba[y_test == 1], label='异常', fill=True, alpha=0.5)
    plt.axvline(best_threshold, color='red', linestyle='--', label='决策阈值')
    plt.xlabel('异常概率')
    plt.ylabel('密度')
    plt.title('预测概率分布')
    plt.legend()
    plt.tight_layout()
    plt.savefig('rf_probability_distribution.png')
    
    plt.close('all')
    
    # 保存模型报告
    with open('rf_model_report.txt', 'w', encoding='utf-8') as f:
        f.write("=== 女胎异常判定模型报告 ===\n\n")
        f.write(f"样本总量: {len(df_female)}\n")
        f.write(f"异常样本比例: {target.mean():.2%}\n\n")
        
        f.write("最佳模型参数:\n")
        for k, v in grid_search.best_params_.items():
            f.write(f"{k}: {v}\n")
        
        f.write("\n特征重要性排序:\n")
        f.write(feature_importances.to_string())
        
        f.write("\n\n测试集性能:\n")
        f.write(f"- AUC: {roc_auc:.4f}\n")
        f.write(f"- 最佳阈值: {best_threshold:.4f}\n")
        
        f.write("\n可视化文件:\n")
        f.write("1. rf_feature_importance.png - 特征重要性\n")
        f.write("2. rf_roc_curve.png - ROC曲线\n")
        f.write("3. rf_confusion_matrix.png - 混淆矩阵\n")
        f.write("4. rf_probability_distribution.png - 概率分布\n")
    
    print("建模完成！结果保存至:")
    print("- rf_feature_importance.png")
    print("- rf_roc_curve.png")
    print("- rf_confusion_matrix.png")
    print("- rf_probability_distribution.png")
    print("- rf_model_report.txt")
except Exception as e:
    print(f"❌ 执行错误: {e}")
    print("📍 错误详情:")
    import traceback
    traceback.print_exc()
    
    # 尝试提供错误解决建议
    if "NoneType" in str(e):
        print("💡 建议: 检查变量是否为None")
    elif "KeyError" in str(e):
        print("💡 建议: 检查字典键或DataFrame列名是否存在")
    elif "IndexError" in str(e):
        print("💡 建议: 检查列表或数组索引是否越界")
    
    raise  # 重新抛出异常以便调试
```

```
# 添加全局错误处理
try:
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    import re
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.model_selection import train_test_split, GridSearchCV
    from sklearn.metrics import roc_curve, auc, confusion_matrix, ConfusionMatrixDisplay
    
    # 设置科学出版级别的可视化风格
    plt.rcParams.update({
        "font.size": 12,
        "font.family": "sans-serif",
        "axes.spines.right": False,
        "axes.spines.top": False,
        "figure.figsize": (10, 6),
        "figure.dpi": 300
    })
    
    # 加载数据
    df = pd.read_excel('附件.xlsx')
    
    # 打印所有列名用于调试
    print("数据集所有列名:")
    for idx, col in enumerate((df.columns if df is not None else [])):
        print(f"{idx}: {col}")
    
    # 定义列名映射规则 (支持模糊匹配)
    column_mapping = {
        "y_concentration": ["Y染色体浓度", "y染色体", "浓度"],
        "trisomy_21": ["21三体结果", "21-三体", "三体21"],
        "trisomy_18": ["18三体结果", "18-三体", "三体18"],
        "trisomy_13": ["13三体结果", "13-三体", "三体13"],
        "x_z": ["X染色体Z值", "X染色体"],
        "gc": ["GC"],  # GC1-GC6会自动匹配
        "bmi": ["孕妇BMI", "BMI"]
    }
    
    # 查找最佳匹配列名
    def find_best_match(patterns, columns):
        """
        使用正则表达式进行模糊列名匹配
        """
        for pattern in patterns:
            regex = re.compile(pattern, re.IGNORECASE)
            matches = [col for col in columns if regex.search(col)]
            if matches:
                return matches[0]
        return None
    
    # 建立列名映射
    actual_columns = {}
    for col_name, patterns in column_mapping.items():
        match = find_best_match(patterns, (df.columns if df is not None else []))
        if match:
            actual_columns[col_name] = match
            print(f"映射 '{col_name}' -> '{match}'")
        else:
            print(f"⚠️ 未找到列: {patterns}")
    
    # 检查关键列是否存在
    required_columns = ["y_concentration", "trisomy_21", "trisomy_18", "trisomy_13", "x_z", "bmi"]
    missing_cols = [col for col in required_columns if col not in actual_columns]
    
    if missing_cols:
        raise KeyError(f"缺少关键列: {', '.join(missing_cols)}")
    
    # 筛选女胎样本 (Y染色体浓度<4%)
    df_female = df[df[actual_columns["y_concentration"]] < 4].copy()
    print(f"女胎样本数量: {len(df_female)}")
    
    # 创建目标变量
    conditions = [
        (df_female[actual_columns["trisomy_21"]] == '高风险'),
        (df_female[actual_columns["trisomy_18"]] == '高风险'),
        (df_female[actual_columns["trisomy_13"]] == '高风险')
    ]
    df_female['异常'] = np.where(np.any(conditions, axis=0), 1, 0)
    print(f"异常样本比例: {df_female['异常'].mean():.2%}")
    
    # 特征工程 - 只使用确认存在的特征
    features = df_female[[
        actual_columns["x_z"],
        actual_columns["bmi"]
    ]]
    
    # 添加GC列（如果存在）
    gc_cols = [col for col in (df.columns if df is not None else []) if "GC" in col][:6]
    if gc_cols:
        print(f"使用GC列: {gc_cols}")
        for col in gc_cols:
            features[col] = df_female[col]
    
    # 添加其他可能存在的特征
    optional_features = {"总读段数", "唯一比对比", "重复比"}
    for feat in optional_features:
        if feat in (df.columns if df is not None else []):
            features[feat] = df_female[feat]
            print(f"添加特征: {feat}")
    
    target = df_female['异常']
    
    # 删除缺失值
    features = features.dropna()
    target = target.loc[features.index]
    print(f"最终样本量: {len(features)}")
    print(f"使用特征: {', '.join((features.columns if features is not None else []))}")
    
    # 划分训练集和测试集
    X_train, X_test, y_train, y_test = train_test_split(
        features, target, test_size=0.3, random_state=42, stratify=target
    )
    
    # 随机森林模型
    rf = RandomForestClassifier(random_state=42)
    
    # 简化的参数网格（加快计算）
    param_grid = {
        'n_estimators': [100, 150],
        'max_depth': [None, 10],
        'min_samples_split': [2, 5],
        'class_weight': [None, 'balanced']
    }
    
    # 网格搜索
    grid_search = GridSearchCV(
        estimator=rf,
        param_grid=param_grid,
        cv=3,  # 减少交叉验证折数
        scoring='f1',
        n_jobs=-1,
        verbose=1
    )
    grid_search.fit(X_train, y_train)
    
    # 最佳模型
    best_rf = grid_search.best_estimator_
    print(f"最佳参数: {grid_search.best_params_}")
    print(f"训练集F1分数: {grid_search.best_score_:.4f}")
    
    # 预测概率
    y_proba = best_rf.predict_proba(X_test)[:, 1]
    
    # ROC曲线和最佳阈值
    fpr, tpr, thresholds = roc_curve(y_test, y_proba)
    roc_auc = auc(fpr, tpr)
    
    # 最佳阈值（最大化Youden指数）
    j_scores = tpr - fpr
    best_idx = np.argmax(j_scores)
    best_threshold = thresholds[best_idx]
    
    y_pred = (y_proba >= best_threshold).astype(int)
    
    # 特征重要性
    feature_importances = pd.Series(
        best_rf.feature_importances_,
        index=(features.columns if features is not None else [])
    ).sort_values(ascending=False)
    
    # === 可视化 ===
    
    # 1. 特征重要性
    plt.figure()
    feature_importances.plot(kind='barh', color='steelblue')
    plt.title('随机森林特征重要性')
    plt.xlabel('重要性得分')
    plt.tight_layout()
    plt.savefig('rf_feature_importance.png')
    
    # 2. ROC曲线
    plt.figure()
    plt.plot(fpr, tpr, color='darkorange', lw=2, 
             label=f'ROC曲线 (AUC = {roc_auc:.2f})')
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.scatter(fpr[best_idx], tpr[best_idx], marker='o', color='red', 
                label=f'最佳阈值={best_threshold:.2f}')
    plt.xlabel('假阳性率')
    plt.ylabel('真阳性率')
    plt.title('ROC曲线')
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.savefig('rf_roc_curve.png')
    
    # 3. 混淆矩阵
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)
    plt.figure()
    disp.plot(cmap='Blues', values_format='d')
    plt.title('混淆矩阵')
    plt.tight_layout()
    plt.savefig('rf_confusion_matrix.png')
    
    # 4. 概率分布
    plt.figure()
    sns.kdeplot(y_proba[y_test == 0], label='正常', fill=True, alpha=0.5)
    sns.kdeplot(y_proba[y_test == 1], label='异常', fill=True, alpha=0.5)
    plt.axvline(best_threshold, color='red', linestyle='--', label='决策阈值')
    plt.xlabel('异常概率')
    plt.ylabel('密度')
    plt.title('预测概率分布')
    plt.legend()
    plt.tight_layout()
    plt.savefig('rf_probability_distribution.png')
    
    plt.close('all')
    
    # 保存模型报告
    with open('rf_model_report.txt', 'w', encoding='utf-8') as f:
        f.write("=== 女胎异常判定模型报告 ===\n\n")
        f.write(f"样本总量: {len(df_female)}\n")
        f.write(f"异常样本比例: {target.mean():.2%}\n\n")
        
        f.write("使用的特征:\n")
        for feat in (features.columns if features is not None else []):
            f.write(f"- {feat}\n")
        
        f.write("\n最佳模型参数:\n")
        for k, v in grid_search.best_params_.items():
            f.write(f"{k}: {v}\n")
        
        f.write("\n特征重要性排序:\n")
        f.write(feature_importances.to_string())
        
        f.write("\n\n测试集性能:\n")
        f.write(f"- AUC: {roc_auc:.4f}\n")
        f.write(f"- 最佳阈值: {best_threshold:.4f}\n")
        
        f.write("\n可视化文件:\n")
        f.write("1. rf_feature_importance.png - 特征重要性\n")
        f.write("2. rf_roc_curve.png - ROC曲线\n")
        f.write("3. rf_confusion_matrix.png - 混淆矩阵\n")
        f.write("4. rf_probability_distribution.png - 概率分布\n")
    
    print("建模完成！结果保存至:")
    print("- rf_feature_importance.png")
    print("- rf_roc_curve.png")
    print("- rf_confusion_matrix.png")
    print("- rf_probability_distribution.png")
    print("- rf_model_report.txt")
except Exception as e:
    print(f"❌ 执行错误: {e}")
    print("📍 错误详情:")
    import traceback
    traceback.print_exc()
    
    # 尝试提供错误解决建议
    if "NoneType" in str(e):
        print("💡 建议: 检查变量是否为None")
    elif "KeyError" in str(e):
        print("💡 建议: 检查字典键或DataFrame列名是否存在")
    elif "IndexError" in str(e):
        print("💡 建议: 检查列表或数组索引是否越界")
    
    raise  # 重新抛出异常以便调试
```

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import roc_curve, auc, confusion_matrix, ConfusionMatrixDisplay

# 设置科学出版级别的可视化风格
plt.rcParams.update({
    "font.size": 12,
    "font.family": "sans-serif",
    "axes.spines.right": False,
    "axes.spines.top": False,
    "figure.figsize": (10, 6),
    "figure.dpi": 300
})

# 加载数据
df = pd.read_excel('附件.xlsx')
print(f"数据集形状: {df.shape}")

# 打印所有列名用于调试
print("\n数据集所有列名:")
for idx, col in enumerate((df.columns if df is not None else [])):
    print(f"{idx}: {col}")

# 定义灵活的列名映射规则
feature_map = {
    "y_concentration": ["Y染色体浓度", ".*[Yy]染色体.*", ".*浓度.*"],
    "x_z": ["X染色体Z值", ".*[Xx]染色体.*", ".*Z值.*"],
    "gc": ["GC1", "GC2", "GC3", "GC4", "GC5", "GC6", ".*GC.*"],
    "bmi": ["孕妇BMI", ".*BMI.*"],
    "total_reads": ["总读段数", ".*读段.*"],
    "unique_align": ["唯一比对比", ".*比对比.*"],
    "dup_rate": ["重复比", ".*重复比.*"],
    "trisomy_21": ["21三体结果", "21-三体结果", ".*21.*三体.*", ".*21.*风险.*"],
    "trisomy_18": ["18三体结果", "18-三体结果", ".*18.*三体.*", ".*18.*风险.*"],
    "trisomy_13": ["13三体结果", "13-三体结果", ".*13.*三体.*", ".*13.*风险.*"]
}

# 查找最佳匹配列名
def find_matches(pattern_list, columns):
    """
    使用正则表达式进行模糊列名匹配
    """
    matches = []
    for pattern in pattern_list:
        try:
            regex = re.compile(pattern, re.IGNORECASE)
            found = [col for col in columns if regex.search(col)]
            if found:
                matches.extend(found)
        except Exception as e:
            print(f"模式编译错误: {pattern} - {e}")
    return list(set(matches))  # 去重

# 建立列名映射
actual_columns = {}
missing = []

for col_name, patterns in feature_map.items():
    matches = find_matches(patterns, (df.columns if df is not None else []))
    if matches:
        # 优先选择精确匹配
        exact_matches = [m for m in matches if m in patterns]
        selected = exact_matches[0] if exact_matches else matches[0]
        actual_columns[col_name] = selected
        print(f"映射 {col_name} -> {selected}")
    else:
        print(f"⚠️ 未找到匹配列: {col_name} ({patterns})")
        missing.append(col_name)

# 检查关键列是否存在
required_columns = ["y_concentration", "x_z", "bmi"]
critical_missing = [col for col in required_columns if col not in actual_columns]

if critical_missing:
    raise KeyError(f"缺少关键列: {', '.join(critical_missing)}")

# 筛选女胎样本 (Y染色体浓度<4%)
if "y_concentration" in actual_columns:
    y_col = actual_columns["y_concentration"]
    df_female = df[df[y_col] < 4].copy()
    print(f"女胎样本数量: {len(df_female)}")
else:
    raise KeyError("Y染色体浓度列不存在")

# 创建目标变量 - 使用替代列
if all(col in actual_columns for col in ["trisomy_21", "trisomy_18", "trisomy_13"]):
    conditions = [
        (df_female[actual_columns["trisomy_21"]] == '高风险'),
        (df_female[actual_columns["trisomy_18"]] == '高风险'),
        (df_female[actual_columns["trisomy_13"]] == '高风险')
    ]
    df_female['异常'] = np.where(np.any(conditions, axis=0), 1, 0)
    print(f"使用三体结果列创建目标变量")
elif "21三体风险" in (df.columns if df is not None else []):
    # 使用替代列
    print(f"使用替代列 '21三体风险' 创建目标变量")
    df_female['异常'] = np.where(df_female['21三体风险'] == '高风险', 1, 0)
else:
    # 最后尝试 - 使用染色体Z值
    print(f"⚠️ 无法找到三体结果列，使用染色体Z值作为替代")
    z_threshold = 3  # 假设Z值>3表示高风险
    df_female['异常'] = np.where(
        (df_female['21三体Z值'] > z_threshold) |
        (df_female['18三体Z值'] > z_threshold) |
        (df_female['13三体Z值'] > z_threshold),
        1, 0
    )

print(f"异常样本比例: {df_female['异常'].mean():.2%}")

# 特征工程 - 只使用确认存在的特征
features = pd.DataFrame()

# 添加核心特征
for feature in ["x_z", "bmi"]:
    if feature in actual_columns:
        features[actual_columns[feature]] = df_female[actual_columns[feature]]

# 添加GC列（如果存在）
gc_matches = find_matches([".*GC.*"], (df.columns if df is not None else []))
if gc_matches:
    print(f"使用GC列: {gc_matches[:min(6, len(gc_matches))]}")
    for col in gc_matches[:6]:
        features[col] = df_female[col]

# 添加其他可能存在的特征
optional_features = {"总读段数", "唯一比对比", "重复比", "21三体Z值", "18三体Z值", "13三体Z值"}
for feat in optional_features:
    if feat in (df.columns if df is not None else []):
        features[feat] = df_female[feat]
        print(f"添加特征: {feat}")

# 确保特征不为空
if features.empty:
    raise ValueError("没有可用的特征列")

target = df_female['异常']

# 删除缺失值
features = features.dropna()
target = target.loc[features.index]
print(f"最终样本量: {len(features)}")
print(f"使用特征: {', '.join((features.columns if features is not None else []))}")

# 划分训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(
    features, target, test_size=0.3, random_state=42, stratify=target
)

# 随机森林模型 - 简化版本
rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    class_weight='balanced'
)

# 训练模型
rf.fit(X_train, y_train)

# 预测概率
y_proba = rf.predict_proba(X_test)[:, 1]

# ROC曲线和最佳阈值
fpr, tpr, thresholds = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)

# 最佳阈值（最大化Youden指数）
j_scores = tpr - fpr
best_idx = np.argmax(j_scores)
best_threshold = thresholds[best_idx]

y_pred = (y_proba >= best_threshold).astype(int)

# 特征重要性
feature_importances = pd.Series(
    rf.feature_importances_,
    index=(features.columns if features is not None else [])
).sort_values(ascending=False)

# === 可视化 ===

# 1. 特征重要性
plt.figure()
feature_importances.plot(kind='barh', color='steelblue')
plt.title('随机森林特征重要性')
plt.xlabel('重要性得分')
plt.tight_layout()
plt.savefig('rf_feature_importance.png')

# 2. ROC曲线
plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, 
         label=f'ROC曲线 (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.scatter(fpr[best_idx], tpr[best_idx], marker='o', color='red', 
            label=f'最佳阈值={best_threshold:.2f}')
plt.xlabel('假阳性率')
plt.ylabel('真阳性率')
plt.title('ROC曲线')
plt.legend(loc='lower right')
plt.tight_layout()
plt.savefig('rf_roc_curve.png')

# 3. 混淆矩阵
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
plt.figure()
disp.plot(cmap='Blues', values_format='d')
plt.title('混淆矩阵')
plt.tight_layout()
plt.savefig('rf_confusion_matrix.png')

# 4. 概率分布
plt.figure()
sns.kdeplot(y_proba[y_test == 0], label='正常', fill=True, alpha=0.5)
sns.kdeplot(y_proba[y_test == 1], label='异常', fill=True, alpha=0.5)
plt.axvline(best_threshold, color='red', linestyle='--', label='决策阈值')
plt.xlabel('异常概率')
plt.ylabel('密度')
plt.title('预测概率分布')
plt.legend()
plt.tight_layout()
plt.savefig('rf_probability_distribution.png')

plt.close('all')

# 保存模型报告
with open('rf_model_report.txt', 'w', encoding='utf-8') as f:
    f.write("=== 女胎异常判定模型报告 ===\n\n")
    f.write(f"样本总量: {len(df_female)}\n")
    f.write(f"异常样本比例: {target.mean():.2%}\n\n")
    
    f.write("使用的特征:\n")
    for feat in (features.columns if features is not None else []):
        f.write(f"- {feat}\n")
    
    f.write("\n特征重要性排序:\n")
    f.write(feature_importances.to_string())
    
    f.write("\n\n测试集性能:\n")
    f.write(f"- AUC: {roc_auc:.4f}\n")
    f.write(f"- 最佳阈值: {best_threshold:.4f}\n")
    
    f.write("\n可视化文件:\n")
    f.write("1. rf_feature_importance.png - 特征重要性\n")
    f.write("2. rf_roc_curve.png - ROC曲线\n")
    f.write("3. rf_confusion_matrix.png - 混淆矩阵\n")
    f.write("4. rf_probability_distribution.png - 概率分布\n")

print("建模完成！结果保存至:")
print("- rf_feature_importance.png")
print("- rf_roc_curve.png")
print("- rf_confusion_matrix.png")
print("- rf_probability_distribution.png")
print("- rf_model_report.txt")
```

```
# 添加全局错误处理
try:
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.model_selection import train_test_split
    from sklearn.metrics import roc_curve, auc, confusion_matrix, classification_report, ConfusionMatrixDisplay
    from sklearn.utils import resample
    
    # 设置科学出版级别的可视化风格
    plt.rcParams.update({
        "font.size": 12,
        "font.family": "sans-serif",
        "axes.spines.right": False,
        "axes.spines.top": False,
        "figure.figsize": (10, 6),
        "figure.dpi": 300,
        "savefig.bbox": "tight"
    })
    sns.set_style("white")
    
    # 加载数据
    df = pd.read_excel('附件.xlsx')
    print(f"数据集形状: {df.shape}")
    
    # 打印所有列名用于参考
    print("\n数据集列名:")
    for idx, col in enumerate((df.columns if df is not None else [])):
        print(f"{idx}: {col}")
    
    # 筛选女胎样本 (Y染色体浓度<4%)
    if 'Y染色体浓度' in (df.columns if df is not None else []):
        df_female = df[df['Y染色体浓度'] < 4].copy()
        print(f"女胎样本数量: {len(df_female)}")
    else:
        # 尝试找到Y染色体浓度列
        y_cols = [col for col in (df.columns if df is not None else []) if "Y" in col or "y" in col or "染色体" in col]
        if y_cols:
            df_female = df[df[y_cols[0]] < 4].copy()
            print(f"使用替代列 '{y_cols[0]}' 筛选女胎样本: {len(df_female)}")
        else:
            raise KeyError("找不到Y染色体浓度列")
    
    # 创建目标变量 - 使用存在的风险列
    if '21三体风险' in (df.columns if df is not None else []):
        df_female['异常'] = np.where(df_female['21三体风险'] == '高风险', 1, 0)
        print(f"使用'21三体风险'创建目标变量")
    elif '21-三体风险' in (df.columns if df is not None else []):
        df_female['异常'] = np.where(df_female['21-三体风险'] == '高风险', 1, 0)
        print(f"使用'21-三体风险'创建目标变量")
    else:
        # 最后尝试使用染色体Z值
        z_cols = [col for col in (df.columns if df is not None else []) if "Z值" in col]
        if z_cols:
            df_female['异常'] = np.where((df_female[z_cols] > 3).any(axis=1), 1, 0)
            print(f"使用染色体Z值创建目标变量")
        else:
            raise KeyError("无法创建目标变量")
    
    print(f"异常样本比例: {df_female['异常'].mean():.2%}")
    
    # 特征工程 - 使用所有数值型特征
    numeric_features = df_female.select_dtypes(include=[np.number]).columns.tolist()
    # 移除目标变量如果包含
    if '异常' in numeric_features:
        numeric_features.remove('异常')
        
    features = df_female[numeric_features]
    target = df_female['异常']
    
    # 删除缺失值
    features = features.dropna()
    target = target.loc[features.index]
    print(f"最终样本量: {len(features)}")
    print(f"使用特征({len((features.columns if features is not None else []))}个): {', '.join((features.columns if features is not None else []).tolist())}")
    
    # 划分训练集和测试集
    X_train, X_test, y_train, y_test = train_test_split(
        features, target, test_size=0.3, random_state=42, stratify=target
    )
    
    # 随机森林模型
    rf = RandomForestClassifier(
        n_estimators=100,
        max_depth=10,
        random_state=42,
        class_weight='balanced'
    )
    
    # 训练模型
    rf.fit(X_train, y_train)
    
    # 预测概率
    y_proba = rf.predict_proba(X_test)[:, 1]
    
    # 找到最佳阈值（最大化Youden指数）
    fpr, tpr, thresholds = roc_curve(y_test, y_proba)
    j_scores = tpr - fpr
    best_idx = np.argmax(j_scores)
    best_threshold = thresholds[best_idx]
    print(f"最佳阈值: {best_threshold:.4f}")
    
    # === 敏感性分析 ===
    print("\n=== 敏感性分析 ===")
    
    # 1. 阈值扰动分析 (±10%)
    thresholds_to_test = np.linspace(best_threshold * 0.9, best_threshold * 1.1, 21)
    results = []
    
    for thresh in thresholds_to_test:
        y_pred = (y_proba >= thresh).astype(int)
        report = classification_report(y_test, y_pred, output_dict=True)
        
        results.append({
            "threshold": thresh,
            "f1": report['weighted avg']['f1-score'],
            "precision": report['weighted avg']['precision'],
            "recall": report['weighted avg']['recall'],
            "accuracy": report['accuracy']
        })
    
    # 转换为DataFrame
    sens_df = pd.DataFrame(results)
    
    # 2. Bootstrap重采样评估置信区间
    n_bootstraps = 100
    bootstrap_results = []
    
    for i in range(n_bootstraps):
        # 重采样
        X_res, y_res = resample(X_test, y_test, random_state=i)
        y_proba_res = rf.predict_proba(X_res)[:, 1]
        
        # 使用最佳阈值
        y_pred_res = (y_proba_res >= best_threshold).astype(int)
        report = classification_report(y_res, y_pred_res, output_dict=True)
        
        bootstrap_results.append({
            "f1": report['weighted avg']['f1-score'],
            "precision": report['weighted avg']['precision'],
            "recall": report['weighted avg']['recall'],
            "accuracy": report['accuracy']
        })
    
    # 计算置信区间
    bootstrap_df = pd.DataFrame(bootstrap_results)
    conf_intervals = bootstrap_df.quantile([0.025, 0.975])
    
    # === 可视化 ===
    
    # 1. 阈值敏感性分析
    plt.figure()
    plt.plot(sens_df['threshold'], sens_df['f1'], 'o-', label='F1分数')
    plt.plot(sens_df['threshold'], sens_df['accuracy'], 's-', label='准确率')
    plt.axvline(best_threshold, color='r', linestyle='--', label='最佳阈值')
    plt.xlabel('分类阈值')
    plt.ylabel('性能指标')
    plt.title('分类阈值敏感性分析')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig('threshold_sensitivity.png')
    
    # 2. Bootstrap置信区间
    metrics = ['f1', 'precision', 'recall', 'accuracy']
    metric_names = ['F1分数', '精确率', '召回率', '准确率']
    
    plt.figure(figsize=(10, 6))
    for i, metric in enumerate(metrics):
        mean_val = bootstrap_df[metric].mean()
        lower = conf_intervals.loc[0.025, metric]
        upper = conf_intervals.loc[0.975, metric]
        
        plt.errorbar(i, mean_val, yerr=[[mean_val - lower], [upper - mean_val]], 
                     fmt='o', capsize=5, label=metric_names[i])
    
    plt.xticks(range(len(metrics)), metric_names)
    plt.ylabel('指标值')
    plt.title('Bootstrap置信区间 (95%)')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig('bootstrap_confidence.png')
    
    # 3. 特征重要性
    feature_importances = pd.Series(
        rf.feature_importances_,
        index=(features.columns if features is not None else [])
    ).sort_values(ascending=False)
    
    plt.figure()
    feature_importances.head(10).plot(kind='barh', color='steelblue')
    plt.title('Top 10特征重要性')
    plt.xlabel('重要性得分')
    plt.tight_layout()
    plt.savefig('feature_importance.png')
    
    # 4. ROC曲线
    roc_auc = auc(fpr, tpr)
    plt.figure()
    plt.plot(fpr, tpr, color='darkorange', lw=2, 
             label=f'ROC曲线 (AUC = {roc_auc:.2f})')
    plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
    plt.scatter(fpr[best_idx], tpr[best_idx], marker='o', color='red', 
                label=f'最佳阈值={best_threshold:.2f}')
    plt.xlabel('假阳性率')
    plt.ylabel('真阳性率')
    plt.title('ROC曲线')
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.savefig('roc_curve.png')
    
    plt.close('all')
    
    # 保存分析报告
    with open('sensitivity_report.txt', 'w', encoding='utf-8') as f:
        f.write("=== 模型敏感性分析报告 ===\n\n")
        f.write(f"样本总量: {len(df_female)}\n")
        f.write(f"异常样本比例: {target.mean():.2%}\n\n")
        
        f.write("最佳阈值: {:.4f}\n\n".format(best_threshold))
        
        f.write("阈值敏感性分析 (±10%):\n")
        f.write(sens_df.round(4).to_string() + "\n\n")
        
        f.write("Bootstrap置信区间 (95%):\n")
        for metric in metrics:
            mean_val = bootstrap_df[metric].mean()
            lower = conf_intervals.loc[0.025, metric]
            upper = conf_intervals.loc[0.975, metric]
            f.write(f"{metric}: {mean_val:.4f} ({lower:.4f} - {upper:.4f})\n")
        
        f.write("\n可视化文件:\n")
        f.write("1. threshold_sensitivity.png - 阈值敏感性分析\n")
        f.write("2. bootstrap_confidence.png - 置信区间\n")
        f.write("3. feature_importance.png - 特征重要性\n")
        f.write("4. roc_curve.png - ROC曲线\n")
    
    print("分析完成！结果保存至:")
    print("- threshold_sensitivity.png")
    print("- bootstrap_confidence.png")
    print("- feature_importance.png")
    print("- roc_curve.png")
    print("- sensitivity_report.txt")
except Exception as e:
    print(f"❌ 执行错误: {e}")
    print("📍 错误详情:")
    import traceback
    traceback.print_exc()
    
    # 尝试提供错误解决建议
    if "NoneType" in str(e):
        print("💡 建议: 检查变量是否为None")
    elif "KeyError" in str(e):
        print("💡 建议: 检查字典键或DataFrame列名是否存在")
    elif "IndexError" in str(e):
        print("💡 建议: 检查列表或数组索引是否越界")
    
    raise  # 重新抛出异常以便调试
```
